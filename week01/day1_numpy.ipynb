## Machine learning
- học máy là một lĩnh vực của AI, là quá trình máy tự học dựa theo những dữ liệu, những quy tắc, những thuật toán, từ đó đưa ra dự đoán mà không cần lập trình thủ công các nguyên tắc
- Ví dụ một hệ thống phân loại email thành spam và thư rác: Lập trình thông thường sẽ dựa vào những quy tắc như là trúng thưởng, có các đường dẫn lạ, hoặc là người gửi nằm trong blacklist, thì khi spammer thay đổi các nguyên tắc đó thì có thể dễ dàng vượt qua filter
- Còn khi lập trình sử dụng machine learning, bạn cung cấp khoảng 1000 mẫu về thư rác và spam, máy sẽ học được các phân loại và gán nhãn các trường hợp nghi ngờ và đưa ra dự đoán cho những email tiếp theo
- Có 3 loại học máy: 
    + học có giám sát: dựa trên dữ liệu có sẵn và các dữ liệu đã được gán nhãn đúng => dự đoán giá nhà, phân loại email, dự đoán sv trượt môn
    + học không giám sát: dựa trên dữ liệu kh có sẵn nhãn đáp án, mô hình sẽ đi tìm cấu trúc và quy tắc của nhóm dữ liệu => chia khách hàng thành các nhóm có hành vi tương tự, phát hiện các nhóm có bài viết cùng chủ đề, phân tích dữ liệu người dùng
    + học tăng cường: một tác nhân tương tác với môi trường, nhận thưởng hoặc phạt học cách chọn hành động => AI học cách chơi game, robot học cách di chuyển, tối ưu chiến lược điều khiển trong môi trường mô phỏng

## Numpy 
- numpy là một thư viện của python, chuyên dùng để xử lý dữ liệu số, tính toán, thực hiện các phép toán đại số tuyến tính 1 cách hiệu quả

```python
## 1 import numpy với tên np
import numpy as np 
```

```python
## 2 in phiên bản numpy và cấu hình build
print(np.__version__)
np.show_config()
```

```python
## 3 tạo vector toàn số 0, kích thước 10
z = np.zeros(10)
print(z)
```

```python
## 4 tính dung luowgj byte của một mảng bất kì
z =np.zeros((10,10))
print(z.size)
print(z.itemsize)
print("%d bytes" % (z.size * z.itemsize))
```

```python
## 7 tạo vector chứa các số từ 10 đến 49
z = np.arange(10,50)
print(z)
```

```python
## 8 đảo ngược vector z
z = np.arange(10,50)
z=z[::-1]
print(z)
```

```python
## 9 tao ma tran 3x3 chua cac so tu 0 den 8
z = np.arange(9).reshape(3,3)
z=z[::-1,::-1]
print(z)
```

```python
## 10 tim chi so cac phan tu khac 0 trong [1,2,0,0,4,0]
z = np.array([1,2,0,0,4,0])
print(np.nonzero(z))
```

```python
## 11 tao ma tran don vi 3x3
z = np.eye(3)
print(z)
```

```python
## 12 tao ma tran don vi 3x3 ngau nhien
z = np.random.random((3,3))
print(z)
```

```python
## 13 tao ma tran 10x10 ngau nhien va tim gia tri lon nhat va nho nhat
z = np.random.random((10,10))
print(z)
print(np.max(z))
print(np.min(z))
```
